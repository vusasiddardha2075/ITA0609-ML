```python
import numpy as np

a = np.array([10, 25, 30, 15, 40])

result = a[a > 20]

print(result)
```

```python
import numpy as np

a = np.array([10, 25, 30, 15, 40])

result = a[(a > 20) & (a < 40)]

print(result)
```

```python
import numpy as np

a = np.array([40, 10, 30, 20, 50])

ascending = np.sort(a)
descending = np.sort(a)[::-1]

print("Ascending order:", ascending)
print("Descending order:", descending)
```

```python
import numpy as np

a = np.array([10, 20, 30, 40, 50])

print("Sum:", np.sum(a))
print("Average:", np.mean(a))
print("Maximum:", np.max(a))
print("Minimum:", np.min(a))
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3, 4, 5])
y = np.array([10, 20, 15, 30, 25])

plt.plot(x, y)
plt.title("Line Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3, 4])
y = np.array([20, 35, 30, 40])

plt.bar(x, y)
plt.title("Bar Chart")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

marks = np.array([45, 50, 55, 60, 65, 70, 70, 75, 80, 85, 90, 95])

plt.hist(marks, bins=5)
plt.title("Histogram")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([10, 20, 30, 40, 50])
y = np.array([15, 25, 35, 45, 55])

plt.scatter(x, y)
plt.title("Scatter Plot")
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

marks = np.array([30, 25, 20, 25])
subjects = ["Maths", "Physics", "Chemistry", "English"]

plt.pie(marks, labels=subjects, autopct="%1.1f%%")
plt.title("Subject Marks")
plt.show()
```
