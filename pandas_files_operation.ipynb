```python
import pandas as pd
import numpy as np
data = {
    "Name": ["siddhu", "venky", "ravi", "sudheer"],
    "Marks": [92, np.nan, 88, np.nan],
    "Age": [20, 21, np.nan, 22]
}
df = pd.DataFrame(data)
print("Original Data:")
print(df)
result = df.dropna()
print("\nAfter removing missing values:")
print(result)
```

```python
import pandas as pd
import numpy as np
data = {
    "Name": ["siddhu", "venky", "ravi", "sudheer"],
    "Marks": [86, np.nan, 90, np.nan]
}
df = pd.DataFrame(data)
print("Original Data:")
print(df)
average = df["Marks"].mean()
df["Marks"] = df["Marks"].fillna(average)
print("\nAfter filling missing values:")
print(df)
```

```python
import pandas as pd
data = {
    "Name": ["siddhu", "venky", "ravi", "sudheer"],
    "Department": ["ECE", "CSE", "ECE", "CSE", "ECE"],
    "Marks": [85, 70, 90, 65, 80]
}
df = pd.DataFrame(data)
print("Original Data:")
print(df)
result = df.groupby("Department")["Marks"].mean()
print("\nAverage marks by department:")
print(result)
```

```python
import pandas as pd
import numpy as np
data = {
    "Name": ["siddhu", "venky", "ravi", "sudheer"],
    "Marks": [85, np.nan, 90, np.nan],
    "Age": [20, 21, np.nan, 22]
}
df = pd.DataFrame(data)
print(df)
print("\nMissing values:")
print(df.isnull())
print("\nCount of missing values:")
print(df.isnull().sum())
```
